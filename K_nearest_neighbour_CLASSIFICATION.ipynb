#👥 K_nearest_neighbour Classification
    K-Nearest Neighbors (KNN) is a non-parametric, lazy learning algorithm.

👉 It does not learn a model.
👉 It stores the data and decides at prediction time.


##🧠 How KNN Classification Works

1.   Choose K (number of neighbors)
2.   Compute distance between test point & all training points
3.   Pick K nearest points
4. Assign class by majority vote


#📏 Distance Metrics (FORMULAS)


*   Euclidean Distance (Most Common)
*   Manhattan Distance



```python
# import libraries
import numpy as np
import pandas as pd
```

```python
#load dataset
from sklearn.datasets import load_breast_cancer
data = load_breast_cancer()
```

```python
data.data
```

```python
data.feature_names
```

```python
data.target
```

```python
data.target_names
```

```python
# create dtaframe
df = pd.DataFrame(np.c_[data.data, data.target], columns=[list(data.feature_names)+['target']])
df.head()
```

```python
df.tail()
```

```python
df.shape
```

```python
# Split Data

X = df.iloc[:, 0:-1]
y = df.iloc[:, -1]
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=2020)

print('Shape of X_train = ', X_train.shape)
print('Shape of y_train = ', y_train.shape)
print('Shape of X_test = ', X_test.shape)
print('Shape of y_test = ', y_test.shape)
```

```python
from sklearn.neighbors import KNeighborsClassifier

classifier = KNeighborsClassifier(n_neighbors=5)
classifier.fit(X_train, y_train)

classifier.score(X_test, y_test)
```

```python
## Predict Cancer

patient1 = [17.99, 10.38, 122.8, 1001.0, 0.1184, 0.2776, 0.3001, 0.1471, 0.2419, 0.07871, 1.095, 0.9053, 8.589, 153.4, 0.006399, 0.04904, 0.05373, 0.01587, 0.03003,
 0.006193, 25.38, 17.33, 184.6, 2019.0, 0.1622, 0.6656, 0.7119, 0.2654, 0.4601, 0.1189]

patient1 = np.array([patient1])
```

```python
patient1
```

```python
classifier.predict(patient1)
```

```python
pred = classifier.predict(patient1)
```

```python

if pred[0] == 0:
  print('Patient has Cancer (malignant tumor)')
else:
  print('Patient has no Cancer (malignant benign)')
```
